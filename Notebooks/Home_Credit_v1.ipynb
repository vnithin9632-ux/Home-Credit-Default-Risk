```python
missing = train.isna().mean()

train.drop(columns=missing[missing > 0.40].index, inplace=True)
```

```python
train.head()
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer()
```

```python
imputer.fit(train.select_dtypes("number"))
```

```python
train_imputed = imputer.transform(train.select_dtypes("number"))
```

```python
train_imputed
```

```python
train_imputed = pd.DataFrame(train_imputed, columns=imputer.feature_names_in_)
train_imputed
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
encoder = OrdinalEncoder()
```

```python
encoder.fit(train.select_dtypes("object"))
```

```python
train_encoded = encoder.transform(train.select_dtypes("object"))
```

```python
train_encoded = pd.DataFrame(train_encoded, columns=encoder.feature_names_in_)
```

```python
train_encoded
```

```python
train_encoded.isna().sum()
```

```python
cat_imputer = SimpleImputer(strategy='median')
```

```python
cat_imputer.fit(train_encoded)
```

```python
cat_imputer = pd.DataFrame(cat_imputer.transform(train_encoded), columns=cat_imputer.feature_names_in_)
```

```python
cat_imputer
```

```python
cat_imputer.isna().sum()
```

```python
completed_train = pd.concat((train_imputed, cat_imputer),axis=1)
completed_train
```

```python
completed_train.isna().sum().sort_values()
```

```python
completed_train["SK_ID_CURR"].unique()
```

```python
X,y = completed_train.drop("TARGET",axis=1), completed_train["TARGET"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, random_state=42, shuffle=True,train_size=0.8)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier(
    n_estimators=500,
    max_depth=15,
    min_samples_split=10,
    min_samples_leaf=5,
    max_features="sqrt",
    class_weight="balanced",
    n_jobs=-1,
    random_state=42
)
```

```python
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import precision_score, recall_score,roc_auc_score,auc, roc_curve, precision_recall_curve
```

```python
y_prob = model.predict_proba(X_test)[:,1]
```

```python
y_prob
```

```python
y_pred = (y_prob >= 0.5).astype(int)
```

```python
precision_score(y_test, y_pred)
```

```python
y_pred
```

```python
model.score(X_test,y_test)
```

```python
roc_auc_score(y_test, y_prob)
```

```python
fpr, tpr, thresholds = roc_curve(y_test, y_prob)
```

```python
auc(fpr,tpr)
```

```python
recall_score(y_test, y_pred)
```

```python
precision,recall, thresholds = precision_recall_curve(y_test, y_prob)
```

```python
plt.plot(recall, precision)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.grid()
plt.show()
```

```python
plt.plot(thresholds, precision[:-1], label="Precision")
plt.plot(thresholds, recall[:-1], label="Recall")

plt.xlabel("Threshold")
plt.ylabel("Score")
plt.title("Precision / Recall vs Threshold")
plt.legend()
plt.grid()
plt.show()
```
